```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

## Create embeddings:

```python
sentences = [
    "I love dogs",
    "I like puppies",
    "I hate traffic"
]

embeddings = model.encode(sentences)

print(embeddings.shape)
```

## Compare sentence similarity

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(embeddings)

print(similarity)
```

## mini semantic search demo

```python
documents = [
    "Dogs are loyal pets",
    "Puppies are very cute",
    "Traffic jams are frustrating",
    "Cats love sleeping",
    "Machine learning is amazing"
]
```

```python
# Create embeddings:
doc_embeddings = model.encode(documents)
```

```python
# Now search query:

query = "I love cute dogs"
query_embedding = model.encode([query])
```

```python
#Similarity:

scores = cosine_similarity(query_embedding, doc_embeddings)

print(scores)
```

## show best match automatically

```python
best_index = scores.argmax()

print("Best Match:", documents[best_index])
print("Score:", scores[0][best_index])
```
