## Step 2 Imports

```python
from transformers import pipeline
```

## Step 3 Load text generation model

```python
generator = pipeline(
    "text-generation",
    model="distilgpt2",
    
)
```

## Step 4 Ask question

What these do:

do_sample=True
→ allows randomness instead of repeating same highest-probability token
temperature=0.7
→ balanced creativity
top_k=50
→ pick from top 50 probable next words
top_p=0.95
→ nucleus sampling
repetition_penalty=1.2
→ punishes repeating same words
max_new_tokens
→ cleaner than max_length

```python
prompt = "Explain AEM"

result = generator(
    prompt,
    max_new_tokens=80,
    do_sample=True,
    temperature=0.7,
    top_k=50,
    top_p=0.95,
    repetition_penalty=1.2
)

print(result[0]["generated_text"])
```
