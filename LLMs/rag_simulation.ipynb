## Step 1 Create fake company knowledge base

```python
docs = [
    "Office timings are 9 AM to 6 PM.",
    "Employees get 20 annual leave days.",
    "Health insurance covers spouse and children.",
    "Work from home is allowed twice a week."
]
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

## Step 2 Convert docs to embeddings
text docs → meaning vectors

```python
doc_embeddings = model.encode(docs)
```

## Step 3 User asks question

```python
query = "How many leave days do employees get?"
query_embedding = model.encode([query])
```

## Step 4 Find best matching doc

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
scores = cosine_similarity(query_embedding, doc_embeddings)

best_index = scores.argmax()

print("Best matching doc:")
print(docs[best_index])
```

## Step 5 Simulate generation

```python
answer = docs[best_index]

print("AI Answer:", answer)
```
