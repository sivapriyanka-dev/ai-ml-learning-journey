## Step 1 Imports

```python
from transformers import pipeline
```

## Step 2 Load pretrained sentiment model

```python
classifier = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)
```

## Step 3 Test

```python
result = classifier("This movie was absolutely amazing. I loved every moment.")
print(result)
```

```python
result = classifier("This movie was terrible and a complete waste of time.")
print(result)
```

```python
print(classifier("I absolutely love this product."))
print(classifier("Worst experience ever. Totally disappointed."))
print(classifier("The movie was not bad at all."))
print(classifier("The food was okay, nothing special."))
```

## Step 4 Compare with old models

```python
texts = [
    "This movie was not good.",
    "This movie was not bad.",
    "I expected better.",
    "Absolutely fantastic performance!",
    "Terrible acting but great visuals."
]

for t in texts:
    print(t)
    print(classifier(t))
    print("------")
```
